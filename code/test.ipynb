```python
import pandas as pd
import numpy as np
import laspy
import open3d as o3d


path_flo = "C:/Users/flori/Desktop/pa-1_data/zurich_1.las"
path_flo_2 = "C:/Users/flori/Desktop/pa-1_data/kantonal_zurich_2.laz"
path_nicola = "C:/Users/Nicola/Desktop/Jona_las.las"

las = laspy.read(path_flo_2)
df = pd.DataFrame({
    "X": np.array(las.x),
    "Y": np.array(las.y),
    "Z": np.array(las.z),
    "Intensity": np.array(las.intensity),
    "Classification": np.array(las.classification)
})

# Show the first few rows
print(df.head())


# Convert LAS data to NumPy array
points = np.vstack((las.x, las.y, las.z)).transpose()
```

```python
# visualization of the pint cloud based on heights

# Create Open3D point cloud object
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)

# Visualize
o3d.visualization.draw_geometries([pcd])
```

```python
## for swisstopo labels

# Extract classification labels
labels = las.classification

# Define a color mapping for different classes
color_map = {
    1: [0.5, 0.5, 0.5],  # Gray for Unclassified
    2: [0.7, 0.4, 0.1],  # Brown for Ground
    3: [0.0, 1.0, 0.0],  # Green for Vegetation
    6: [0.8, 0.0, 0.0],  # Red for Buildings
    9: [0.0, 0.0, 1.0],  # Blue for Water
    17: [1.0, 1.0, 0.0],  # Yellow for Bridges
}

# Apply colors based on classification labels
colors = np.array([color_map.get(label, [1, 1, 1]) for label in labels])  # Default white for unknown labels

# Create Open3D point cloud object
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)
pcd.colors = o3d.utility.Vector3dVector(colors)

# Visualize the point cloud
o3d.visualization.draw_geometries([pcd], window_name="Colored Point Cloud")
```
